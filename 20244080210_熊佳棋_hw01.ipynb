欧几里得算法

```python
# 1. 实现 gcd 函数
def gcd(m, n):
    while n != 0:
        m, n = n, m % n
    return m

# 2. 评估数字对
print("GCD(2018, 2019) =", gcd(2018, 2019))
print("GCD(1200, 300) =", gcd(1200, 300))
print("GCD(5040, 60) =", gcd(5040, 60))

# 3. 测试负数情况
print("\n--- 负数情况测试 ---")
print("GCD(-12, 8) =", gcd(-12, 8))
print("GCD(-12, -8) =", gcd(-12, -8))
```

关于负数参数的回答：
当我们向 gcd 传入负数时，函数依然会返回正的最大公约数。这是因为在 Python 中，取模运算 % 对于负数的结果总是非负数。
这种行为是有意义的：在数学定义中，最大公约数通常被定义为正数，所以该函数正确地遵循了数学定义。如果不希望处理负数，可以在函数开头对输入取绝对值，或者抛出异常。

近似欧拉数e

```python
import math

# 1. 基于极限公式近似 e
def euler_limit(n):
    return (1 + 1/n)**n

# 2. 基于无穷级数前 n 项之和近似 e
def euler_infinite_sum(n):
    if n == 0:
        return 0.0
    total = 0.0
    factorial = 1.0
    for k in range(n):
        if k > 0:
            factorial *= k
        total += 1.0 / factorial
    return total

# 3. 使用 while 循环，达到指定精度 epsilon
def euler_approx(epsilon):
    true_e = math.exp(1)
    total = 0.0
    factorial = 1.0
    k = 0
    while abs(total - true_e) > epsilon:
        if k > 0:
            factorial *= k
        total += 1.0 / factorial
        k += 1
    return total

# 4. 打印从 1 到 n 的近似值表
def print_euler_sum_table(n):
    for k in range(1, n+1):
        print(f"n={k}: {euler_infinite_sum(k)}")

# --- 测试代码 ---
print("euler_limit(10) =", euler_limit(10))
print("euler_infinite_sum(4) =", euler_infinite_sum(4)) # 预期 2.666...
print("euler_approx(0.001) =", euler_approx(0.001))
print("\n--- 级数表 ---")
print_euler_sum_table(5)
```

测试整数属性

```python
# 1. 判断是否是 2 的幂
def is_power_of_2(n):
    if n <= 0:
        return False
    while n > 1:
        if n % 2 != 0:
            return False
        n = n // 2
    return True

# 2. 判断 n 是否是 b 的幂
def is_power(b, n):
    if b <= 1 or n <= 0: # 基础合法性检查
        return False
    if n == 1: # 任何数的 0 次方都是 1
        return True
    while n > 1:
        if n % b != 0:
            return False
        n = n // b
    return True

# --- 测试代码 ---
print("--- is_power_of_2 测试 ---")
print("2 是 2 的幂:", is_power_of_2(2))
print("3 是 2 的幂:", is_power_of_2(3))
print("8 是 2 的幂:", is_power_of_2(8))
print("10 是 2 的幂:", is_power_of_2(10))

print("\n--- is_power 测试 ---")
print("3^4=81，81 是 3 的幂:", is_power(3, 81))
print("3^4=81，80 是 3 的幂:", is_power(3, 80))
print("5^1=5，5 是 5 的幂:", is_power(5, 5))
```
