实验2：函数、条件、循环与递归

任务1：编写score_level(score)函数

```python
def score_level(score):
    # 检查是否超出 0-100 范围
    if score < 0 or score > 100:
        raise ValueError(f"成绩 {score} 超出 0~100 范围！")
    
    # 条件判断
    if 90 <= score <= 100:
        return "优秀"
    elif 80 <= score < 90:
        return "良好"
    elif 60 <= score < 80:
        return "及格"
    else:
        return "不及格"

# 测试函数
print("95分:", score_level(95))
print("85分:", score_level(85))
print("70分:", score_level(70))
print("50分:", score_level(50))

# 测试异常捕获
try:
    score_level(105)
except ValueError as e:
    print("捕获到异常:", e)
```

任务2：使用for和while计算1到100的和

```python
# 1. 使用 for 循环
sum_for = 0
for i in range(1, 101):
    sum_for += i
print("for 循环结果:", sum_for)

# 2. 使用 while 循环
sum_while = 0
j = 1
while j <= 101 - 1:  # 即 j <= 100
    sum_while += j
    j += 1
print("while 循环结果:", sum_while)

print("两者是否相等:", sum_for == sum_while)
```

任务3：递归版和迭代版阶乘函数

```python
# 迭代版
def factorial_iter(n):
    if n < 0:
        raise ValueError("输入必须是非负整数")
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

# 递归版
def factorial_rec(n):
    if n < 0:
        raise ValueError("输入必须是非负整数")
    if n == 0:
        return 1
    return n * factorial_rec(n - 1)

# 比较测试 n = 0, 1, 5, 10
test_ns = [0, 1, 5, 10]
for n in test_ns:
    print(f"n={n}: 迭代版={factorial_iter(n)}, 递归版={factorial_rec(n)}")
```

任务4：summary(values)函数与元组解包

```python
def summary(values):
    if not values:  # 检查列表是否为空
        raise ValueError("列表不能为空")
    
    min_val = min(values)
    max_val = max(values)
    avg_val = sum(values) / len(values)
    
    # 返回元组
    return (min_val, max_val, avg_val)

# 测试并解包
data = [85, 58, 92, 76, 60, 89]
minimum, maximum, average = summary(data)  # 元组解包
print(f"最小值: {minimum}, 最大值: {maximum}, 平均值: {average:.2f}")
```

任务5：sum_all(*args)可变参数函数

```python
def sum_all(*args):
    # *args 会将传入的参数打包成一个元组
    return sum(args)

# 测试情况
print("无参数:", sum_all())
print("单个参数:", sum_all(10))
print("多个参数:", sum_all(1, 2, 3, 4, 5))
```

实验4：文件读写、异常处理与模块

任务1~3：创建目录、写入文件、编写解析函数

```python
import os
import pickle # 提前导入后面要用

# --- 任务 1：路径与目录 ---
print("当前工作目录:", os.getcwd())
data_dir = os.path.join(os.getcwd(), "exp01_data")
os.makedirs(data_dir, exist_ok=True) # exist_ok=True 防止重复创建报错
print(f"数据目录已就绪: {data_dir}")

# --- 任务 2：写入数据到 students.txt ---
file_path = os.path.join(data_dir, "students.txt")
data_content = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""

with open(file_path, "w", encoding="utf-8") as f:
    f.write(data_content)
print("文件写入成功！")

# --- 任务 3：编写 parse_student(line) 函数 ---
def parse_student(line):
    line = line.strip()
    if not line: # 遇到空行直接抛错
        raise ValueError("空行")
    
    parts = line.split(",")
    if len(parts) != 2:
        raise ValueError(f"字段数量错误，期望2个，实际{len(parts)}个")
    
    name, score_str = parts[0].strip(), parts[1].strip()
    if not name:
        raise ValueError("姓名为空")
    
    try:
        score = int(score_str)
    except ValueError:
        raise ValueError(f"成绩不是合法整数: '{score_str}'")
        
    if not (0 <= score <= 100):
        raise ValueError(f"成绩 {score} 超出 0~100 范围")
        
    return {"name": name, "score": score}

print("parse_student 函数定义完成")
```

任务4~5：处理异常、统计并写入report.txt

```python
# --- 任务 4：按行处理文件，捕获异常并收集有效数据 ---
valid_records = []
invalid_count = 0

print("开始处理文件记录：")
with open(file_path, "r", encoding="utf-8") as f:
    next(f) # 跳过表头（name,score）
    for line_no, line in enumerate(f, start=2): # 行号从2开始（1是表头）
        try:
            record = parse_student(line)
            valid_records.append(record)
        except ValueError as e:
            invalid_count += 1
            print(f"-> 第 {line_no} 行无效 (原始内容: '{line.strip()}')，原因: {e}")

print(f"\n处理完成！有效记录: {len(valid_records)} 条，无效记录: {invalid_count} 条。")

# --- 任务 5：计算统计量并写入 report.txt ---
report_path = os.path.join(data_dir, "report.txt")

if valid_records:
    valid_scores = [r["score"] for r in valid_records]
    avg_score = sum(valid_scores) / len(valid_scores)
    pass_count = sum(1 for s in valid_scores if s >= 60)
    pass_rate = pass_count / len(valid_scores)
    
    report_text = (
        f"=== 实验4 成绩统计报告 ===\n"
        f"有效记录数: {len(valid_records)}\n"
        f"无效记录数: {invalid_count}\n"
        f"平均分: {avg_score:.2f}\n"
        f"及格率: {pass_rate:.1%}\n"
    )
    
    # 演示 "w" 模式 (覆盖写入)
    with open(report_path, "w", encoding="utf-8") as f:
        f.write(report_text)
    print("\n已使用 'w' 模式写入 report.txt (覆盖原有内容)")
    
    # 演示 "a" 模式 (追加写入)
    with open(report_path, "a", encoding="utf-8") as f:
        f.write("--- 补充说明 ---\n")
        f.write("此报告由 Python 脚本自动生成。\n")
    print("已使用 'a' 模式追加内容到 report.txt")
    
    # 验证读取
    print("\n--- report.txt 当前内容 ---")
    with open(report_path, "r", encoding="utf-8") as f:
        print(f.read())
```

实验5：类、继承与对象复制

任务1~2：定义Student类并创建对象

```python
# --- 任务 1：定义 Student 类 ---
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"Student(name='{self.name}', score={self.score})"

print("Student 类定义完成。")

# --- 任务 2：为有效记录创建对象并输出 ---
print("\n--- 学生对象信息 ---")
# 这里的 valid_records 是我们在实验4中生成的有效数据列表
student_objects = []
for record in valid_records:
    # 使用字典里的数据初始化对象
    stu = Student(record["name"], record["score"])
    student_objects.append(stu)
    
    # 输出对象信息和及格状态
    status = "及格" if stu.is_passed() else "不及格"
    print(f"{stu} -> 状态: {status}")
```

任务3：继承GraduateStudent类

```python
# --- 任务 3：定义继承自 Student 的 GraduateStudent 类 ---
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        # 调用父类的初始化方法
        super().__init__(name, score)
        self.research_topic = research_topic

    # 重写 __str__ 方法
    def __str__(self):
        # 可以调用父类的 __str__ 或者自己重写全部
        return f"GraduateStudent(name='{self.name}', score={self.score}, topic='{self.research_topic}')"

print("GraduateStudent 类定义完成。\n")

# 测试继承
grad_stu = GraduateStudent("张三", 95, "人工智能")
print("对象信息:", grad_stu)
print("是否及格:", grad_stu.is_passed()) # 从父类继承的方法
```

任务4：对象复制

```python
import copy

print("--- 对象复制对比 ---")

# 每次比较都从新建的原始对象开始
def create_original():
    return {"name": "Alice", "scores": [85, 90]}

# 1. 直接赋值
print("\n1. 直接赋值 (Assignment):")
original_1 = create_original()
assigned = original_1
assigned["scores"].append(100) # 修改内部列表
print("原对象:", original_1)
print("赋值后:", assigned)

# 2. 浅拷贝 (copy.copy)
print("\n2. 浅拷贝 (Shallow copy):")
original_2 = create_original()
shallow = copy.copy(original_2)
shallow["scores"].append(100) # 修改内部列表
print("原对象:", original_2)
print("浅拷贝后:", shallow)

# 3. 深拷贝 (copy.deepcopy)
print("\n3. 深拷贝 (Deep copy):")
original_3 = create_original()
deep = copy.deepcopy(original_3)
deep["scores"].append(100) # 修改内部列表
print("原对象:", original_3)
print("深拷贝后:", deep)
```
